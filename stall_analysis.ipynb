### **Problem Statement**
Wanjiku owns a fresh produce stall in Gikomba Market and wants to understand how her business is performing. This analysis uses 14 days of sales data and inventory information to evaluate sales performance, growth trends and stock requirements.



```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## **Sales summary**
This task calculates total sales, average sales, maximum sales, minimum sales and identifies the best and worst performing days.

```python
total_sales = sum(daily_sales)

average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)

lowest_sales = min(daily_sales)
```

```python
best_position = daily_sales.index(highest_sales)

best_day = days[best_position]

best_week = 1 if best_position < 7 else 2
```

```python
worst_position = daily_sales.index(lowest_sales)

worst_day = days[worst_position]

worst_week = 1 if worst_position < 7 else 2
```

```python
print("Sales Summary")
print(f"Total Sales: KES {total_sales:,}")
print(f"Average Sales: KES {average_sales:,.2f}")
print(f"Highest Sales: KES {highest_sales:,}")
print(f"Lowest Sales: KES {lowest_sales:,}")
print(f"Best Day: {best_day} (Week {best_week})")
print(f"Worst Day: {worst_day} (Week {worst_week})")
```

## **Week-on-Week Growth**
This task compares sales in Week 1 and Week 2 using a reusable function.

```python
def week_totals(sales):
    """
    Return week 1 total and week 2 total sales.
    """

    week1_total = sum(sales[:7])

    week2_total = sum(sales[7:])

    return week1_total, week2_total
```

```python
week1_total, week2_total = week_totals(daily_sales)
```

```python
week1_total, week2_total
```

```python
growth_pct = ((week2_total - week1_total)
              / week1_total) * 100
```

```python
print(f"Week 1 Total: KES {week1_total:,}")
print(f"Week 2 Total: KES {week2_total:,}")
print(f"Growth: {growth_pct:.2f}%")
```

## **Day Classification**
This task classifies daily performance as Strong, Normal or Weak.

```python
def classify_day(amount, average):
    """
    Classify sales performance.
    """

    if amount > average * 1.2:
        return "Strong"

    elif amount < average * 0.8:
        return "Weak"

    else:
        return "Normal"
```

```python
classification_counts = {}
```

```python
for sale in daily_sales:

    label = classify_day(
        sale,
        average_sales
    )

    classification_counts[label] = (
        classification_counts.get(label, 0)
        + 1
    )
```

```python
print(classification_counts)
```

## **Weekend Vs Weekday Sales**
Compare weekday performance against weekend performance.

```python
weekday_sales = []

weekend_sales = []
```

```python
for day, sale in zip(days, daily_sales):

    if day in ["Mon",
               "Tue",
               "Wed",
               "Thu",
               "Fri"]:

        weekday_sales.append(sale)

    else:

        weekend_sales.append(sale)
```

```python
weekday_avg = (
    sum(weekday_sales)
    / len(weekday_sales)
)

weekend_avg = (
    sum(weekend_sales)
    / len(weekend_sales)
)
```

```python
difference_pct = (
    (weekend_avg - weekday_avg)
    / weekday_avg
) * 100
```

```python
print(f"Weekday Average: KES {weekday_avg:,.2f}")

print(f"Weekend Average: KES {weekend_avg:,.2f}")

print(f"Difference: {difference_pct:.2f}%")
```

## **Inventory Analysis **
Identify products requiring restocking and calculate gross margins.

```python
def reorder_products(stock_data):
    """
    Return products that need reordering.
    """

    to_reorder = []

    for product, info in stock_data.items():

        if info["qty"] <= info["reorder"]:

            to_reorder.append(product)

    return to_reorder
```

```python
products_to_reorder = reorder_products(stock)

print(products_to_reorder)
```

(price - cost) / price

```python
for product, info in stock.items():

    margin = (
        (info["price"] - info["cost"])
        / info["price"]
    ) * 100

    print(
        f"{product}: "
        f"{margin:.1f}%"
    )
```

## **Stall Performance Report**

```python
print("=" * 60)
print("STALL PERFORMANCE REPORT")
print("=" * 60)

print(f"Total Sales: KES {total_sales:,.0f}")

print(f"Average Daily Sales: KES {average_sales:,.2f}")

print(
    f"Best Day: {best_day} "
    f"(Week {best_week}) - "
    f"KES {highest_sales:,.0f}"
)

print(
    f"Worst Day: {worst_day} "
    f"(Week {worst_week}) - "
    f"KES {lowest_sales:,.0f}"
)

print(f"Week 1 Sales: KES {week1_total:,.0f}")

print(f"Week 2 Sales: KES {week2_total:,.0f}")

print(f"Growth Rate: {growth_pct:.2f}%")

print(f"Weekday Average: KES {weekday_avg:,.2f}")

print(f"Weekend Average: KES {weekend_avg:,.2f}")

print()

print(
    "Products To Reorder:"
)

print(", ".join(products_to_reorder))
```
